1. Konsep Singkat

Misalnya kita ingin mengetahui apakah rata-rata nilai siswa setelah belajar dengan metode baru berbeda dari nilai standar 75.

Dataset sederhana:

```python
scores = [78, 80, 77, 85, 82, 79, 81, 84, 76, 83]
```

Kita gunakan one-sample t-test karena ingin membandingkan rata-rata sampel dengan satu nilai standar.

A. Two-tailed Test

Null Hypothesis dan Alternative Hypothesis

Null Hypothesis (H0):
Rata-rata nilai siswa sama dengan 75.

Alternative Hypothesis (H1):
Rata-rata nilai siswa tidak sama dengan 75.

Secara sederhana:

H0: μ = 75
H1: μ ≠ 75

Two-tailed test digunakan ketika kita ingin mengecek apakah rata-rata berbeda, bisa lebih tinggi atau lebih rendah.

```python
# Import library
import numpy as np
from scipy import stats

# Dataset sederhana
scores = [78, 80, 77, 85, 82, 79, 81, 84, 76, 83]

# Nilai standar yang ingin diuji
standard_value = 75

# One-sample t-test
t_statistic, p_value = stats.ttest_1samp(scores, standard_value)

# Menampilkan hasil
print("Dataset:", scores)
print("Mean:", np.mean(scores))
print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Tingkat signifikansi
alpha = 0.05

# Keputusan
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The average score is significantly different from 75.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: The average score is not significantly different from 75.")
```

Penjelasan Output

Jika p-value < 0.05, maka H0 ditolak. Artinya rata-rata nilai siswa berbeda secara signifikan dari 75.

Jika p-value >= 0.05, maka H0 tidak ditolak. Artinya belum cukup bukti bahwa rata-rata nilai siswa berbeda dari 75.

B. One-tailed Test

Sekarang kita ingin mengetahui apakah rata-rata nilai siswa lebih tinggi dari 75.

Null Hypothesis dan Alternative Hypothesis

Null Hypothesis (H0):
Rata-rata nilai siswa kurang dari atau sama dengan 75.

Alternative Hypothesis (H1):
Rata-rata nilai siswa lebih dari 75.

Secara sederhana:

H0: μ ≤ 75
H1: μ > 75

One-tailed test digunakan ketika arah pengujian sudah jelas, misalnya hanya ingin menguji apakah nilai lebih tinggi.



```python
# Import library
import numpy as np
from scipy import stats

# Dataset sederhana
scores = [78, 80, 77, 85, 82, 79, 81, 84, 76, 83]

# Nilai standar yang ingin diuji
standard_value = 75

# One-sample t-test
t_statistic, p_value_two_tailed = stats.ttest_1samp(scores, standard_value)

# Karena scipy menghasilkan two-tailed p-value,
# untuk one-tailed test kita bagi p-value dengan 2
p_value_one_tailed = p_value_two_tailed / 2

print("Dataset:", scores)
print("Mean:", np.mean(scores))
print("T-statistic:", t_statistic)
print("Two-tailed P-value:", p_value_two_tailed)
print("One-tailed P-value:", p_value_one_tailed)

# Tingkat signifikansi
alpha = 0.05

# Keputusan untuk H1: mean > 75
if t_statistic > 0 and p_value_one_tailed < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The average score is significantly higher than 75.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not enough evidence that the average score is higher than 75.")
```

C. Two-sample Test

Sekarang kita bandingkan dua kelompok siswa.

Kelompok A menggunakan metode belajar lama.
Kelompok B menggunakan metode belajar baru.



```python
method_old = [72, 74, 70, 73, 75, 71, 74, 72]
method_new = [78, 80, 77, 85, 82, 79, 81, 84]
```

Null Hypothesis dan Alternative Hypothesis

Null Hypothesis (H0):
Rata-rata nilai kelompok lama sama dengan kelompok baru.

Alternative Hypothesis (H1):
Rata-rata nilai kelompok lama berbeda dengan kelompok baru.

Secara sederhana:

H0: μ_old = μ_new
H1: μ_old ≠ μ_new



```python
# Import library
import numpy as np
from scipy import stats

# Dataset dua kelompok
method_old = [72, 74, 70, 73, 75, 71, 74, 72]
method_new = [78, 80, 77, 85, 82, 79, 81, 84]

# Independent two-sample t-test
t_statistic, p_value = stats.ttest_ind(method_old, method_new)

print("Old Method Mean:", np.mean(method_old))
print("New Method Mean:", np.mean(method_new))
print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Tingkat signifikansi
alpha = 0.05

# Keputusan
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant difference between the two methods.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is no significant difference between the two methods.")
```

D. Contoh Output Interpretasi

Misalnya hasilnya seperti ini:

Old Method Mean: 72.625
New Method Mean: 80.75
P-value: 0.0001
Decision: Reject H0
Conclusion: There is a significant difference between the two methods.

Artinya, rata-rata nilai siswa dengan metode baru berbeda secara signifikan dari metode lama. Karena rata-rata metode baru lebih tinggi, metode baru dapat dianggap memberikan hasil yang lebih baik berdasarkan dataset sederhana tersebut.

| Jenis Test        | Tujuan                                                       | Hypothesis            |
| ----------------- | ------------------------------------------------------------ | --------------------- |
| One-sample t-test | Membandingkan rata-rata sampel dengan satu nilai standar     | H0: μ = nilai standar |
| One-tailed test   | Menguji arah tertentu, misalnya lebih besar atau lebih kecil | H1: μ > nilai standar |
| Two-tailed test   | Menguji apakah ada perbedaan tanpa menentukan arah           | H1: μ ≠ nilai standar |
| Two-sample t-test | Membandingkan rata-rata dua kelompok                         | H0: μ1 = μ2           |
